```python
#python program that writes each line in reverse order
with open("next.txt","r") as f:
  data =f.readlines()
  for i in data:
    print(i[::-1])
```

```python

#write a program to sort words in a file and put them in another file . the output file should only
#lower case words , so any upper case words are removed
with open("next.txt","r") as f2:
  words = f2.readlines()
words = [w.strip() for w in words if w.strip().islower()]
words.sort()
with open("next.txt","w") as f3:
  for w in words :
    f3.write(w + "\n")
```

modes to open binary files arerb,wb,ab.

```python
import pickle
with open("s1.dat","wb") as f:
  data = (1,2,3,4,5,"hello")
  pickle.dump(data,f)
with open("s1.dat","rb") as f2:
  d=pickle.load(f2)
  print(d)
```
