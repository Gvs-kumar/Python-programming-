```python
#write an program to find duplicates in a list
l=[int(i) for i in input().split()]
s=set(l)
print(len(l)-len(s))
```

```python
#to find no.of numbers having duplicates in a list
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```

```python
#to find how many numbers are having duplicates
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)==1:
    l2.append(i)
print(l2)
```

```python
#to print missing numbers betteen 1 to n
n=int(input())
l=[int(i) for i in input().split()]
l1=set(l)
elements=set(range(1,n+1))
print(elements)
print(elements-l1)
```

```python
#set methods
a={1,2,3,4}
a.add(5) #add
print(a)
#update
a.update([4,5,6])
print(a)
#remove
a.remove(3)
print(a)
#discard
a.discard(99)
print(a)
b={2,3,4,7}
print(a.difference(b))#difference
print(a.union(b))#union
print(a.intersection(b))#intersection
print(a.symmetric_difference(b))#symmetric difference
print(a.pop())#pop
print(a.clear())#clear
```
