```python
#tuple comprehension
t=(int(i) for i in input().split())
print(t)
t1=(i**2 for i in range(1,5))
print(t1)
print(tuple(t))
print(tuple(t1))
```

```python
#ZIP OPERATION
a1=("n1","n2","n3","n4")
a2=(11,22,33,44)
a3=(55,66)
print(tuple(zip(a1,a2)))
print(list(zip(a1,a2)))
print(list(zip(a1,a3)))
```
