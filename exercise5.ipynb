```python
#progrm to reverse a string using loop
def revstring(s):
  x=""
  for i in range(len(s)-1,-1,-1):
    x = x + s[i]
  return x
print(revstring("santhosh"))

```

```python
#palindrome checking
def reverse_number(n):
  rev = 0
  temp = n
  while temp > 0:
    digit = temp % 10
    rev = rev * 10 + digit
    temp = temp // 10
  return rev
num = int(input("Enter a number: "))
reversed_num = reverse_number(num)
if num == reversed_num:
  print("palindrom")
else:
  print("alaa kaadhu raa")
```

```python
#function that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_average = multiple_function(1,2,3,4,5,6,8)
print(f" the sum of numbers is {res_sum}, the count is {res_count},average is {res_average:.2f}")
```
